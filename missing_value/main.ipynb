```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error
```

```python
melb_path_data = "melb_data.csv"

melb_read_data = pd.read_csv(melb_path_data)

melb_read_data.describe()
```

```python
y = melb_read_data.Price


features = ["Rooms","Distance", "Bathroom", "Landsize",  "YearBuilt"]
X = melb_read_data[features]



X_train, X_val, y_train, y_val = train_test_split(X, y, random_state=1)

cols_with_missing = [col for col in X_train.columns
                        if X_train[col].isnull().any()]

reduced_X_train = X_train.drop(cols_with_missing, axis=1)
reduced_X_valid = X_val.drop(cols_with_missing, axis=1)
```

```python
# Imputation

from sklearn.impute import SimpleImputer

my_imputer = SimpleImputer()


imputed_X_train = pd.DataFrame(my_imputer.fit_transform(X_train))
imputed_X_val = pd.DataFrame(my_imputer.transform(X_val))

imputed_X_train.columns = X_train.columns
imputed_X_val.columns = X_val.columns
```

```python
# Extention Imputation 

X_train_plus = X_train.copy()
X_val_plus = X_val.copy()

for col in cols_with_missing:
    X_train_plus[col + '_was_missing'] = X_train_plus[col].isnull()
    X_val_plus[col + '_was_missing'] = X_val_plus[col].isnull()

imputed_X_train_plus = pd.DataFrame(my_imputer.fit_transform(X_train_plus))
imputed_X_val_plus = pd.DataFrame(my_imputer.transform(X_val_plus))

imputed_X_train_plus.columns = X_train_plus.columns
imputed_X_val_plus.columns = X_val_plus.columns
```

```python
model = HistGradientBoostingRegressor(max_iter=200, random_state=1)

model.fit(imputed_X_train_plus, y_train)

val_predict = model.predict(imputed_X_val_plus)

mae = mean_absolute_error(y_val, val_predict)

print(mae)
```
