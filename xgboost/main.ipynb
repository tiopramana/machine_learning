```python
import pandas as pd
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error
```

```python
data_path = "../Project/House_prices/home-data-for-ml-course/train.csv"

read_data = pd.read_csv(data_path)

read_data.head()
```

```python
# Train the model 

y = read_data.SalePrice

houseprices = ['MSSubClass', 'LotFrontage', 'LotArea', 'MoSold', 'YrSold']

X = read_data[houseprices]

X_train, X_valid, y_train, y_valid = train_test_split(X, y, random_state=1)
```

```python
# Fitting model

my_model = XGBRegressor(n_estimators=500, early_stopping_rounds=5, learning_rate=0.5, random_state=1)
# my_model.fit(X_train, y_train)
my_model.fit(X_train, y_train, 
            eval_set  = [(X_valid, y_valid)],
            verbose = False
)

preds = my_model.predict(X_valid)
print(mean_absolute_error(y_valid, preds))
```
