```python
import pandas as pd
from sklearn.metrics import mean_absolute_error
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split


def get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=0)
    model.fit(train_X, train_y)
    prediction_val = model.predict(val_X)
    mae = mean_absolute_error(val_y, prediction_val)
    return (mae)

```

```python

melbroune_data_path = "melb_data.csv"

melbroune_data = pd.read_csv(melbroune_data_path)

melbroune_data.describe()

filtered_melbroune_data = melbroune_data.dropna(axis=0)

y = filtered_melbroune_data.Price

melbroune_features = ['Rooms', 'Bathroom', 'Landsize', 'BuildingArea', 'YearBuilt', 'Lattitude', 'Longtitude']

X = filtered_melbroune_data[melbroune_features]

```

```python
from sklearn.model_selection import train_test_split

train_X, val_x, train_y, val_y = train_test_split(X, y, random_state=0)
```

```python
for max_leaf_nodes in [5,50,500,5000]:
    model = get_mae(max_leaf_nodes, train_X, val_x, train_y, val_y)
    print("Max leaf nodes: %d  \t\t Mean Absolute Error:  %d" %(max_leaf_nodes, model))

# candidate_max_leaf_nodes = [5, 25, 50, 100, 250, 500]
# # Write loop to find the ideal tree size from candidate_max_leaf_nodes
# hasil = {leaf_size: get_mae(leaf_size, train_X, val_X, train_y, val_y) for leaf_size in candidate_max_leaf_nodes}


# # Store the best value of max_leaf_nodes (it will be either 5, 25, 50, 100, 250 or 500)
# best_tree_size = min(hasil, key=hasil.get)
```
